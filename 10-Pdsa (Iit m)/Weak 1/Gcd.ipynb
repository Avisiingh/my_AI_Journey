##### Computing gcd

```python
def gcd(m,n):
    cf = []
    for i in range(1,min(m,n)+1):
        if m % i == 0 and n % i ==0:
            cf.append(i)
    return cf[-1]
```

```python
gcd(8,12)
```

```python
## another way
def gcd(m,n):
    for i in range(1,min(m,n)+1):
        if m % i == 0 and n % i ==0:
            mrcf = i
    return mrcf
```

```python
gcd(8,12)
```
