### Feature Engineering
- Feature engineering is the process of using domain knowledge to extract features from raw data. These features can be used to improve the performance of machine learning algorithms.

#### Main Branches of Feature Engineering

##### 1. **Feature Transformation**

Transforming existing features to make them more useful for a model.

- **Missing Value Imputation**  
  *Filling in missing data so the model can use the feature.*  
  **Example:**  
  If a column "Age" has missing values, you can fill them with the mean age or a special value (like -1).

- **Handling Categorical Features**  
  *Converting text categories into numbers so models can use them.*  
  **Example:**  
  For a "Color" column with values "Red", "Blue", "Green", you can use one-hot encoding to create separate columns for each color.

- **Outlier Detection**  
  *Identifying and handling data points that are very different from others.*  
  **Example:**  
  If most people's income is between $30k and $100k, but one value is $1 million, you might remove or cap this outlier.

- **Feature Scaling**  
  *Making sure all features are on a similar scale so no single feature dominates.*  
  **Example:**  
  Scaling "Age" (0-100) and "Income" (0-100,000) to both be between 0 and 1 using MinMaxScaler.



##### 2. **Feature Construction**

Creating new features from existing data to help the model.

**Example:**  
If you have "Date of Birth", you can create a new feature "Age".  
If you have "Length" and "Width", you can create "Area" = Length × Width.

##### 3. **Feature Selection**

Choosing the most important features and removing irrelevant ones.

**Example:**  
If you have 100 features, but only 10 are useful for predicting the target, you can use feature selection techniques (like correlation, mutual information, or model-based selection) to keep only those 10.

##### 4. **Feature Extraction**

Automatically creating new features from raw data, often using algorithms.

**Example:**  
- In image data, using PCA (Principal Component Analysis) to reduce thousands of pixel values to a few important features.
- In text data, using TF-IDF or word embeddings to convert text into numeric features.
