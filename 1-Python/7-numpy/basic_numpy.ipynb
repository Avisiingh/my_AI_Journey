##### Array Operation

```python
## Arthimatic Operations
a =np.array([10,20,30,40])
b =np.array([1,2,3,4])

c = a- b
print(c)

d = a * b
print(d)
```

```python
farenheit = np.array([0,-10,-5,-15,0])
celsius  =(farenheit-31)*(5/9)
print(celsius)
```

```python
# matrix product
A = np.array([[1,1],[0,1]])
B = np.array([[2,0],[3,4]])
print(A @ B)
```

```python
## Element product
print(A*B)
```
