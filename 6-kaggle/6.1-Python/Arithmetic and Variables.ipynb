Using multiple variables


```python
# Create variables
num_year =4
days_per_year =365
hours_per_day =24
mins_per_hour =60
secs_per_min =60

# Calculate number of seconds in four years
total_sec =num_year* days_per_year*hours_per_day*mins_per_hour*secs_per_min
print(total_sec)
```

```python
# Update to include leap years
days_per_year = 365.25

# Calculate number of seconds in four years
total_secs = num_year* days_per_year*hours_per_day*mins_per_hour*secs_per_min
print(total_secs)
```
