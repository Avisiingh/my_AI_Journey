##### Adding a column in the DataFrame

```python
import pandas as pd
import numpy as np
```

```python
df =pd.DataFrame(np.random.rand(9,4),index=list("abcdefghi"),columns=list("ABCD"))
```

```python
df.head()
```

```python
## adding a column
df["E"]=df["A"]* 100
```

```python
df.head()
```

```python
df["F"] =df["A"] + df["C"]
```

```python
df.head()
```

```python
criteria =df["A"] < 0.2
```

```python
criteria
```

```python
df.loc[criteria,"A"] =0
```

```python
df
```

```python
cities =["mumbai","Delhi","Chennai",
         "Kolkata","Bengalure","Hydrabad",
         "Pune","Ahmedabad","Indore"]
```

```python
df["city"] =cities
```

```python
df
```

```python
df_copy= df.copy()
```

```python
df_copy
```

```python
cities_new=["mumbai","Delhi","Chennai",
        "Kolkata","Bengalure","Hydrabad",
        "Pune","Ahmedabad","Guwahati"]
```

```python
df_copy["new_city"] =cities_new
```

```python
df_copy
```

```python
criteria =df_copy["city"].isin(["Pune","Bengaluru","Hydrabad"])
```
