#### Graoup by

Three stages
- Split - we split dataframe into multiple smaller dataframe based on the values of keys
- Apply - we apply desired aggregation/tranformation on each datframe
- Combine -we combine results from aplly state into a datframe

```python
df=pd.DataFrame({"key":["A","B","C"]*2,
                 "data": range(6)})
```

```python
df
```

```python
df.groupby("key").sum()
```
