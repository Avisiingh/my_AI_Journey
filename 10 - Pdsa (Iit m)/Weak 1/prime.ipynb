##### Checking prime number

```python
def factors(n):
    f1 = []
    for i in range(1,n+1):
        if n % i ==0:
            f1.append(i)
    return f1
```

```python
factors(10)
```

```python
def prime(n):
    return (factors(n) ==[1,n])
```

```python
prime(11)
```

##### Counting prime

```python
## list of prime number upto
def primesupto(m):
    pl = []
    for i in range(1,m+1):
        if prime(i):
            pl.append(i)
    return(pl)
```

```python
primesupto(100)
```

```python
def firstprime(m):
    (count, i ,pl) = (0,1,[])
    while (count < m):
        if prime(i):
            (count,pl) =(count + 1 , pl + [i])
        i = i + 1
    return(pl)
```

```python
firstprime(100)
```

```python
import math
def prime(m):
    (result,i) =(True,2)
    while (result and (i < math.sqrt(m))):
        if m % i ==0:
            return False
        i = i + 1
    return (result)
```

```python
prime(11)
```

```python
def primediffs(n):
    lastprime = 2
    pd = {}
    for i in range(3,  n+ 1):
        if prime(i):
            d = i - lastprime
            if d in pd.keys():
                pd[d] = pd[d] + 1
            else:
                pd[d] =1
    return(pd)
```
