##### Element-wise functions of vectors

```python
## squre 
x =np.arange(1,5)
y = x **2
y
```

```python
## Taking log10
x = np.array([1,10,100,1000,10_000,100_000])
y =np.log10(x)
y
```

```python
## taking log to base e
x = np.array([1,np.e,np.e**2,np.e**3])
y =np.log(x)
y
```

```python
## addition of scalar in each element
x=np.array([1,2,3])
c =5
y = x + c
y
```

```python
## f(x) =5x-2
x =np.array([-1,0,1,2])
y = 5*x - 2
y
```

##### Dot product

```python
x =np.arange(1,5)
y =np.arange(-4,-8,-1)

x @ y
```

```python
x =np.arange(1,5)
y =np.arange(-4,-8,-1)

x.T @ y  ## T = transpose
```

```python
## another way
np.dot(x,y)
```

##### Norms of a vectors

```python
x =np.array([1,2,3])
np.linalg.norm(x)   ## L2 norm
```

```python
x =np.array([1,2,3])
np.linalg.norm(x,ord=1)  ## L1 norm
```
